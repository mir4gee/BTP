# 04 — Few-shot Node Classification (official code)

Runs the official `scripts/1-shot_NC/` and `scripts/5-shot_NC/` scripts on each downstream dataset to reproduce Table 3.

```python
import os
OFFICIAL_DIR = '/content/RiemannGFM_official'
BASE = '/content/drive/MyDrive/RiemannGFM/baseline'
assert os.path.isdir(OFFICIAL_DIR), 'Run 00_setup.ipynb first.'
%cd $OFFICIAL_DIR
os.makedirs(f'{BASE}/results/fewshot', exist_ok=True)
!ls scripts/1-shot_NC scripts/5-shot_NC
```

```python
# 1-shot — iterate over every script in scripts/1-shot_NC/.
import glob
for script in sorted(glob.glob('scripts/1-shot_NC/*.sh')):
    name = os.path.splitext(os.path.basename(script))[0]
    log = f'{BASE}/results/fewshot/1shot_{name}.log'
    print(f'\n=== 1-shot {name} ===')
    !bash $script 2>&1 | tee $log
```

```python
# 5-shot.
for script in sorted(glob.glob('scripts/5-shot_NC/*.sh')):
    name = os.path.splitext(os.path.basename(script))[0]
    log = f'{BASE}/results/fewshot/5shot_{name}.log'
    print(f'\n=== 5-shot {name} ===')
    !bash $script 2>&1 | tee $log
```
